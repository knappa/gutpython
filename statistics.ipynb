# Visualize age/energy

```python
def get_fig(time: int, loc: int, bact_type: str):
    fig, ax = plt.subplots()
    for idx in range(200):
        with h5py.File(data_storage_dir + f"/sample-{str(idx).zfill(4)}.hdf5", "r") as f:
            data = f[str(time)]
            locs = data[bact_type + "_locations"].astype(np.int64)
            mask = np.all(locs == np.array([loc, 0]), axis=1)
            plt.scatter(data[bact_type + "_age"][mask], data[bact_type + "_energy"][mask])
    ax.vlines(np.array([0, 330, 1000 + time]), 0, 130, colors="k")
    ax.hlines(np.array([0, 25, 50, 80, 130]) - 0.5, 0, 1000 + time, colors="k")
```

```python
get_fig(675, 30, "clost")
```
